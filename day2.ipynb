```python
import pandas as pd
import numpy as np

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("all packages added successfully!")
```

```python
#creating a dataframe from dictionary
data = {
    'student_name': ['emon', 'muuko', 'anamika', 'ravi', 'sita'],
    'hours_studied': [5, 3, 7, 2, None],
    'sleep_hours': [7, 5, 8, 4, 6],
    'previous_score': [85, 70, 92, 60, 78],
    'attendance': [90, 75, 95, 60, 85],
    'extracurricular': ['Yes', 'No', 'Yes', 'No', 'Yes'],
    'parent_education': ['bachelor', 'highschool', 'master', 'highschool', 'bachelor']
    
}

df = pd.DataFrame(data)
print("Student Performance DataFrame")
print(df)

print("data info")
print(df.info())

#basic dataframe information
print(f"\nDataFrame Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print("column names:", df.columns.to_list())
print("data types:\n ",df.dtypes)

```
